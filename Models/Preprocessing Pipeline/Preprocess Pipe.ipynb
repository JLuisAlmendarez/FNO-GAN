```python
import sys
import torch
import numpy as np
from pathlib import Path
```

```python
PROJECT_ROOT = Path("/home/joseluisalmendarezgonzalez/Desktop/3_GEO_FNO")
RAW_DATA = PROJECT_ROOT / "Models" / "Dataset" / "Kolmogorov2d_fp32_64x64_N1152_Re1000_T100.pt"
OUTPUT_FILE = PROJECT_ROOT / "Models" / "Dataset" / "snapshots_64x64_use.npy"
```

```python
data = torch.load(RAW_DATA, map_location="cpu", weights_only=False)
vort = data["vorticity"].numpy()                     # (N, T, H, W)
```

```python
vmin = vort.min(axis=(1, 2, 3), keepdims=True)
vmax = vort.max(axis=(1, 2, 3), keepdims=True)
vort_norm = (vort - vmin) / (vmax - vmin + 1e-8)
```

```python
np.save(OUTPUT_FILE, vort_norm)
print(f"Guardado: {OUTPUT_FILE}")
print(f"Shape: {vort_norm.shape}  Min: {vort_norm.min():.4f}  Max: {vort_norm.max():.4f}")
```
